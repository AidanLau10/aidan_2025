## Unit 4: Iteration (17.5 – 22.5 %)  


### Enduring Understanding  
- **CON-2:** Repetition structures enable concise processing of repeated tasks. 

### Essential Knowledge  
- Loop constructs: `while`, `do–while`, `for`; differences in entry/exit conditions.  
- Loop components: initialization, test condition, update expression.  
- Nested loops for multidimensional data traversal.  
- Common loop patterns: accumulation (summing), searching, counting occurrences.

### Suggested Skills  
- Trace loop execution step-by-step with sample indices/values.  
- Convert one loop type to another (e.g., `while` ↔ `for`) to deepen understanding.  
- Implement algorithms: sum of array elements, linear search, frequency counts.

### Tips  
- **Off-by-one vigilance:** Carefully index loops to avoid skipping first/last elements.  
- **Loop selection:** Choose the most readable loop type for each scenario.  
- **Nested visualization:** Draw grids when working with nested loops over 2D data.
