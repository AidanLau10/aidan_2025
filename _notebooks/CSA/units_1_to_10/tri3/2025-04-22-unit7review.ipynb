## Unit 7: ArrayList (2.5 – 7.5 %)  


### Enduring Understanding  
- **IOC-3:** Dynamic lists adapt size, simplifying element addition/removal.

### Essential Knowledge  
- `ArrayList<E>` declaration, import (`java.util.ArrayList`).  
- Core methods: `add()`, `remove()`, `get()`, `set()`, `size()`, `clear()`.  
- Autoboxing/unboxing with wrapper types (`ArrayList<Integer>`). 

### Suggested Skills  
- Manipulate lists: add/remove elements, access by index, iterate both with indices and enhanced `for`.  
- Convert between arrays and `ArrayList`. 

### Tips  
- **When to use:** Prefer `ArrayList` for dynamic collections; arrays for fixed-size, performance-critical tasks.  
- **Edge-case handling:** Removing from empty lists or invalid indices throws exceptions—practice safe checks.  
- **Iteration patterns:** Compare `for (int i…)` vs. `for (E elem : list)` for clarity and safety.
