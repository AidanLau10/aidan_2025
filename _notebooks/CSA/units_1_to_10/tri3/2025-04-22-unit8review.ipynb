## Unit 8: 2D Arrays (7.5 – 10 %)  


### Enduring Understanding  
- **VAR-3:** Nested arrays model grids, facilitating matrix-like computations. 

### Essential Knowledge  
- Declaring and initializing `Type[][]`; rectangular vs. ragged arrays.  
- Nested loop traversal: row-major order; row lengths and index bounds.  
- Typical operations: counting, summing, neighbor checks, pathfinding basics. 

### Suggested Skills  
- Traverse 2D arrays with nested loops, performing per-row and per-column operations.  
- Handle irregular (“ragged”) arrays safely.  
- Implement common grid algorithms: row sums, column sums, diagonal checks. 

### Tips  
- **Grid visualization:** Sketch small matrices to map indices before coding.  
- **Break tasks:** First operate on one row/column, then generalize to nested loops.  
- **Boundary checks:** For neighbor checks, guard against invalid indices.
