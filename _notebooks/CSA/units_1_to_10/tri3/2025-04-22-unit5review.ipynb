## Unit 5: Writing Classes (5 – 7.5 %)  


### Enduring Understanding  
- **IOC-1:** Encapsulation and clear interfaces promote modular, maintainable code. 

### Essential Knowledge  
- Class design: fields (instance variables), methods, constructors.  
- Access modifiers: `public`, `private`; encapsulation principles.  
- The `this` reference; default vs. explicit constructors.  
- Overriding `toString()` and `equals()` for meaningful object representations and comparisons.

### Suggested Skills  
- Sketch UML-like diagrams to plan class attributes and behaviors.  
- Write unit-test “driver” classes with `main` methods to validate each class.  
- Apply consistent naming conventions for classes, methods, fields.

### Tips  
- **Incremental development:** Build and test one method/constructor at a time.  
- **Code readability:** Use clear comments and Javadoc for public methods.  
- **Equality contracts:** Ensure `equals()` and `hashCode()` are coherent when overridden.
