# Homework

Initial:

> Start at node A.
- g(A) = 0
- h(A) = 6
- f(A) = g + h = 0 + 6 = 6

Open list: A
Closed list: (empty)

## Step 1: Expand A (f = 6)
> Neighbors of A:
1. B: 
- g(B) = g(A) + cost(A–B) = 0 + 2 = 2
- h(B) = 4
- f(B) = 2 + 4 = 6

2. C:
- g(C) = 0 + 4 = 4
- h(C) = 2
- f(C) = 4 + 2 = 6

Open list: B (f=6), C (f=6)
Closed list: A


## Step 2: Expand B (f = 6)
(Tie-break: pick B over C arbitrarily)

> Neighbors of B:
1. C (already in open list):
- g through B = 2 + 3 = 5 (worse than existing g(C) = 4), so do not update

2.D:
- g(D) = 2 + 5 = 7
- h(D) = 0
- f(D) = 7 + 0 = 7

Open list: C (f=6), D (f=7)
Closed list: A, B

## Step 3: Expand C (f = 6)
> Neighbors of C:
1. D (already in open list):
- g through C = 4 + 1 = 5 < current g(D) = 7 → update
- g(D) = 5
- h(D) = 0
- f(D) = 5 + 0 = 5

Open list: D (f=5)
Closed list: A, B, C

## Step 4: Expand D (f = 5) → Goal reached
Final Shortest Path: A → C → D
Total cost = g(D) = 5



# Multiple-Choice Answers
1. Which property of a heuristic h(n) guarantees that A* will find the optimal path?
→ B. Admissibility
(Ensures h(n) ≤ true cost to goal → guarantees optimality)

2. Greedy Best‑First Search selects which node to expand next?
→ B. The node with the smallest h(n)
(Greedy = always follows what looks closest to goal)

3. On a 4‑connected grid, which is an admissible heuristic?
→ B. Manhattan distance
(Matches grid movement — up/down/left/right)

4. Drawback of Greedy Best‑First Search vs A*?
→ B. It may expand nodes in non‑optimal order and fail to find the shortest path
(Because it ignores g(n), only follows h(n))
