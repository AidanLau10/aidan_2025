## Unit 9: Inheritance & Polymorphism (5 – 10 %)  


### Enduring Understanding  
- **IOC-4:** Inheritance and dynamic binding enable flexible, extensible designs. 

### Essential Knowledge  
- `extends` keyword; class hierarchies; `super()` constructor calls.  
- Method overriding; `@Override` annotation; dynamic method dispatch at runtime.  
- `instanceof` checks and downcasting (when allowed). 

### Suggested Skills  
- Design superclass/subclass relationships, factoring shared behavior upward.  
- Override methods to customize behavior; test polymorphic calls via superclass references. 

### Tips  
- **Inheritance diagrams:** Draw UML trees to plan class hierarchies.  
- **Override safely:** Always use `@Override` to catch signature mismatches.  
- **Polymorphic tests:** Pass subclass instances to methods expecting superclass types.
