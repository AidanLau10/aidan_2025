## Unit 10: Recursion (5 – 7.5 %)  


### Enduring Understanding  
- **CON-4:** Recursive decomposition can simplify solutions for self-similar problems. 

### Essential Knowledge  
- Define base case vs. recursive case; ensure progression toward base case.  
- Common patterns: factorial, Fibonacci, recursive array/`ArrayList` processing.  
- Call stack behavior; potential stack overflow.

### Suggested Skills  
- Trace small-scale recursive calls, annotating parameter values and return flow.  
- Convert simple recursion to iterative loops to compare approaches.

### Tips  
- **Base-case focus:** Always identify and test the simplest case first.  
- **Call tracing:** Draw recursion trees for sample inputs to visualize flow.  
- **Performance checks:** Recognize exponential blow-up in naive recursion (e.g., Fibonacci) and practice memoization or iterative alternatives.
