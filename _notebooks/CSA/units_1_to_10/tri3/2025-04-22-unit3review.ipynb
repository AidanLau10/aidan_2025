## Unit 3: Boolean Expressions & `if` Statements (15 – 17.5 %)  


### Enduring Understanding  
- **CON-1:** Control flow is governed by boolean evaluations and logical constructs.

### Essential Knowledge  
- Relational operators: `==`, `!=`, `<`, `>`, `<=`, `>=`.  
- Logical operators: `&&`, `||`, `!`; short-circuit evaluation semantics.  
- Boolean expression construction and truth-table analysis.  
- `if`, `if–else`, nested `if`; common pitfalls (e.g., using `=` vs. `==`).  
- Combining relational and logical operators; precedence rules.

### Suggested Skills  
- Simplify complex boolean expressions using De Morgan’s laws.  
- Write and trace nested `if` statements to determine program paths.  
- Test boundary and edge cases (equalities, zero, negative values). 

### Tips  
- **Truth-table practice:** Create tables to verify multi-clause expressions.  
- **Edge-case scenarios:** Specifically test `true/false` for each relational check.  
- **Code reviews:** Locate and correct accidental assignments in conditional statements.
