## Unit 2: Using Objects (5 – 7.5 %)  


### Enduring Understandings  
- **MOD-1:** Object libraries speed development and reduce errors.  
- **VAR-1:** Reference variables hold object addresses or `null`.

### Essential Knowledge  
- Class/object distinction; `new` keyword; constructor signatures and overloading.  
- Calling methods: `void` vs. non-`void`, with/without parameters; return-value usage.  
- `String` API: concatenation, literals, `length()`, `substring()`, `equals()`, immutability.  
- Wrapper classes (`Integer`, `Double`): constructors, constants (`MAX_VALUE`, `MIN_VALUE`), `intValue()`, autoboxing.  
- Static methods (e.g., `Math.random()`, `Math.max()`); `null` reference consequences. 

### Suggested Skills  
- Instantiate objects via various constructors and invoke methods to manipulate state.  
- Write code snippets using `String` and `Math` methods; predict outputs.  
- Recognize and avoid `NullPointerException` by validating references.

### Tips  
- **API flashcards:** Memorize common methods for `String`, `Integer`, `Double`, and `Math`.  
- **Trace calls:** Step through method invocations on paper to follow parameter passing and returns.  
- **Null-check patterns:** Practice pre-call checks (`if (obj != null)`) to guard against exceptions.
