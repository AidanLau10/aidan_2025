## Unit 6: Arrays (10 – 15 %)  


### Enduring Understanding  
- **IOC-2:** Fixed-size arrays organize homogeneous data for indexed access. 

### Essential Knowledge  
- Declaring and initializing one-dimensional arrays; literal vs. `new` syntax.  
- Access via indices; `.length` property; index-out-of-bounds errors.  
- Standard array algorithms: traversal, sum, max/min, linear search, reverse in-place. 

### Suggested Skills  
- Implement and test array-processing loops for common tasks.  
- Handle empty or single-element arrays safely.  
- Use `.length` rather than hard-coding bounds.

### Tips  
- **Bounds checking:** Always validate index before array access.  
- **Algorithm practice:** Write from-scratch implementations of search and sort helpers.  
- **Corner cases:** Test arrays of size 0 and 1 to ensure robustness.
