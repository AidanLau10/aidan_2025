## Unit 1: Primitive Types (2.5 – 5 %)  

### Enduring Understandings  
- **MOD-1:** Leveraging existing, tested classes/methods accelerates reliable programming.  
- **VAR-1:** Variables parameterize algorithms, enabling reuse with different inputs. 

### Essential Knowledge  
- Console output: `System.out.print` vs. `System.out.println` and newline behavior.  
- Primitive types: `int`, `double`, `boolean`; memory model; `final` constants.  
- Arithmetic operators: `+`, `-`, `*`, `/`, `%`; integer vs. floating-point division.  
- Compound assignments (`+=`, `-=`, `*=`, `/=`, `%=`) and increment/decrement (`++`, `--`).  
- Operator precedence, parentheses, and evaluation order.  
- Casting rules: implicit widening, explicit narrowing, and potential runtime errors.

### Suggested Skills  
- Translate problem statements into variable declarations/assignments.  
- Manually evaluate mixed-type expressions, noting truncation or rounding.  
- Identify and correct compile-time errors from invalid casts or mis-typed assignments.

### Tips  
- **Error-hunting drills:** Introduce syntax/type mistakes to practice debugging.  
- **Mini-program practice:** Write small `main` methods to print intermediate expression results.  
- **Division deep dive:** Experiment with different operand types to internalize integer vs. double behavior.
